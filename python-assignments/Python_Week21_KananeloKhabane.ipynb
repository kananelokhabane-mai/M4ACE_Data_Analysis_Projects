```python
def check_balance():
    print (f"\n Your current account balance is: ${account_balance}")
```

```python
def transfer_money():
    global account_balance
    amount = float(input("Enter amount to transfer: $"))
    if amount > account_balance:
        print("Insufficient funds!")
    else:
        account_balance -= amount
        account_no = input("Enter destination account number: ")
        print(f"Successfully transferred ${amount} to Account: {account_no}")
```

```python
def start_ussd_app():
    print("Dialing *111# ...")
    dial_code = input("Enter USSD Code: ")
    if dial_code != "*111#":
        print("Invalid USSD Code. Connection terminated.")
        return
    running = True
    while running:
        print("\n=== WELCOME TO KAYK BANK ===")
        print("1. Check Balance")
        print("2. Transfer Money")
        print("3. Exit")
        
        choice = input("Select an option (1-3): ")
        
        if choice == "1":
            check_balance()
        elif choice == "2":
            transfer_money()
        elif choice == "3":
            print("\n Thank you for banking with KayK Bank. Session closed.")
            running = False # This breaks the loop and closes the app
        else:
            print("Invalid selection! Please try again.")
```

```python
start_ussd_app()
```
